# 01 - Validación de Calidad, Conteos y Verificación de Tablas Gold

```python
dbutils.widgets.text("catalogo", "electrocasa_dev", "Catálogo")
catalogo = dbutils.widgets.get("catalogo")

tablas_silver = ["ventas", "catalogo", "empleados", "resenas", "devoluciones", "sucursales", "tracking_envios"]
tablas_gold = [
    "ventas_sucursal_mes",
    "top_productos",
    "dotacion_mensual",
    "ventas_vendedor",
    "resenas_categoria",
    "despachos_courier"
]

print("=== CONTEOS CAPA SILVER ===")
for t in tablas_silver:
    cnt = spark.table(f"{catalogo}.silver.{t}").count()
    print(f"{catalogo}.silver.{t}: {cnt} filas")

print("\n=== AUDITORÍA DE CUARENTENA ===")
df_cuarentena = spark.table(f"{catalogo}.silver.cuarentena")
df_cuarentena.groupBy("fuente", "motivo_rechazo").count().show(truncate=False)

print("=== CONTEOS Y VALIDACIÓN CAPA GOLD ===")
for t in tablas_gold:
    cnt = spark.table(f"{catalogo}.gold.{t}").count()
    print(f"{catalogo}.gold.{t}: {cnt} filas")
    assert cnt > 0, f"ERROR CRÍTICO: La tabla {catalogo}.gold.{t} está vacía."
```
